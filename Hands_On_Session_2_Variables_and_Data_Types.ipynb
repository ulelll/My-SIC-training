## 1️⃣ Variables & Assignment 📦

Think of a **variable** as a **labeled box** where you store a piece of information. You give the box a name (the variable name), and put something inside it (the value) using the `=` sign.

```
   x = 5
   ┌─────────┐
   │    5    │  <- box labeled "x"
   └─────────┘
```

You can always look inside the box later, or replace what's inside it (reassign it). Let's try it out. 👇


````python
# Basic variable assignment
x = 5
print(x)
````

````python
# Reassigning a variable - the box now holds a new value
x = 5
print("Before reassignment:", x)

x = 10
print("After reassignment:", x)
````

````python
full_name = "rahmat fajri"
````

````python
# Assignment operators - a shortcut for "update the box based on what's already inside"
score = 10
print("Starting score:", score)

# score = score + 5
score += 5   # same as: score = score + 5
print("After += 5:", score)

score -= 3   # same as: score = score - 3
print("After -= 3:", score)

score *= 2   # same as: score = score * 2
print("After *= 2:", score)

score /= 4   # same as: score = score / 4
print("After /= 4:", score)
````

## 2️⃣ Data Types Overview 🔤🔢

Every value in Python has a **type**. The four most common beginner types are:

- 🔢 **`int`** — whole numbers, e.g. `5`, `-3`, `100`
- 🌊 **`float`** — decimal numbers, e.g. `3.14`, `-0.5`
- 🔤 **`str`** — text (strings), written in quotes, e.g. `"hello"`
- ✅ **`bool`** — `True` or `False` values

You can check any value's type using the built-in `type()` function.


````python
# One variable of each type
age = 16                # int
height = 1.65            # float
name = "Priya"           # str
is_student = True        # bool

print(name, "->", type(name))
print(age, "->", type(age))
print(height, "->", type(height))
print(is_student, "->", type(is_student))
````

## 3️⃣ Type Conversion 🔄

Sometimes you need to convert a value from one type to another. Python gives us handy functions for this:

- `int()` → convert to a whole number
- `float()` → convert to a decimal number
- `str()` → convert to text

Let's see some examples, including a **common mistake** beginners make.


````python
# Converting between types
num_str = "25"
num_int = int(num_str)      # string -> int
print(num_int, type(num_int))

price_str = "9.99"
price_float = float(price_str)   # string -> float
print(price_float, type(price_float))

age = 16
age_str = str(age)          # int -> string
print(age_str, type(age_str))
````

### ⚠️ Common Mistake: Mixing Types

You **cannot** directly combine a string and a number with `+`. Python doesn't know if you mean "add them like numbers" or "join them like text," so it raises a `TypeError`.

Run the cell below to see the error in action (this is expected!):


````python
# ❌ Broken example - this will raise a TypeError
age = 25
message = "My age is: " + age
print(message)
````

👆 That error happens because `age` is an `int`, and you can't `+` it directly to a `str`. To fix it, convert `age` to a string first using `str()`:


````python
# ✅ Corrected version - convert age to a string first
age = 25
message = "My age is: " + str(age)
print(message)
````

## 4️⃣ The `input()` Function ⌨️

`input()` lets your program ask the user a question and wait for them to type an answer.

⚠️ **Important:** `input()` **always** returns a **string**, even if the user types a number! If you want to do math with it, you must convert it first using `int()` or `float()`.


````python
# Ask for the user's name and greet them
name = input("What's your name? ")
print("Hello, " + name + "! Welcome to Session 2.")
````

````python
# Ask for a number and convert it before doing math
age_str = input("How old are you? ")
age = int(age_str)   # convert the string to an int before using it in math

next_year_age = age + 1
print("Next year, you will be", next_year_age, "years old.")
````

## 5️⃣ Basic String & Number Operations ➕✖️

Strings and numbers each support their own set of handy operations:

- Strings: `+` (concatenation/joining), `*` (repetition), `len()` (length)
- Numbers: `+ - * /` (basic math), `%` (modulo — the remainder after division)


````python
# String concatenation and repetition
first = "Py"
second = "thon"
full_word = first + second        # joining strings
print(full_word)

cheer = "Go! " * 3                # repeating a string
print(cheer)

print("Length of full_word:", len(full_word))
````

````python
# Number operations
a = 17
b = 5

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)   # division always returns a float
print("a % b =", a % b)   # modulo - the remainder of a / b
````
