## 🚦 2. Apa Itu Control Flow?

**Control flow** adalah cara program menentukan **jalur mana** yang harus dijalankan, berdasarkan suatu **kondisi**.

Analogi paling gampang: **lampu lalu lintas** 🚦
- Kalau lampu **hijau** → jalan
- Kalau lampu **merah** → berhenti

Alurnya seperti ini:

```
MULAI
  ↓
Kondisi terpenuhi?
  ↓ Ya                    ↓ Tidak
Jalankan Jalur A         Jalankan Jalur B
  ↓
LANJUT KODE BERIKUTNYA
```

Di Python, kita mengekspresikan ini dengan pernyataan **`if`**.

## 🧩 3. Pernyataan `if`

### 📖 Penjelasan
Pernyataan `if` menjalankan satu blok kode **HANYA JIKA** kondisinya bernilai `True`.

```python
if kondisi:
    # blok kode ini hanya jalan kalau kondisi True
    ...
```

**Anatomi `if`:**
| Bagian | Contoh | Keterangan |
|---|---|---|
| `if` | `if` | kata kunci pembuka |
| kondisi | `nilai > 60` | harus menghasilkan `True`/`False` |
| `:` | `:` | titik dua, **wajib ada** |
| indentasi | `    print(...)` | blok kode, biasanya 4 spasi |

⚠️ **Perhatian:** Python sangat memperhatikan **indentasi** (spasi di awal baris). Kalau indentasi salah, kode akan error atau berjalan tidak sesuai harapan.


````python
# Contoh: pernyataan if
nilai = 80

if nilai > 60:
    print("Kamu lulus!")

print("Program selesai.")  # baris ini tetap jalan walau kondisi False
````

````python
# Contoh: pernyataan if
nilai = 80

# nilai >= 80
if nilai >= 80:
    print("Kamu lulus!")

print("Program selesai.")  # baris ini tetap jalan walau kondisi False
````

## ⚖️ 4. Pernyataan `if-else`

### 📖 Penjelasan
Bagaimana jika kita ingin program menjalankan hal **lain** saat kondisi salah? Gunakan `else`.

```python
if kondisi:
    # jalan jika kondisi True
    ...
else:
    # jalan jika kondisi False
    ...
```

🔑 **Ingat:** salah satu blok **PASTI** berjalan — tidak pernah keduanya, dan tidak pernah tidak ada.


````python
# Contoh: if-else
suhu = 18

if suhu >= 25:
    print("Pakai kaos")
else:
    print("Pakai jaket")
````

## 🪜 5. Pernyataan `if-elif-else`

### 📖 Penjelasan
Bagaimana jika ada **lebih dari dua** kemungkinan? Gunakan `elif` (singkatan dari *"else if"*) untuk menambah cabang.

```python
if kondisi_1:
    ...
elif kondisi_2:
    ...
elif kondisi_3:
    ...
else:
    ...
```

Python mengecek dari **atas ke bawah**, dan **berhenti** di cabang pertama yang bernilai `True`. Cabang setelahnya tidak akan dicek lagi!


````python
# Contoh: if-elif-else untuk nilai huruf
nilai = 72

if nilai >= 90:
    print("Nilai A")
elif nilai >= 75:
    print("Nilai B")
elif nilai >= 60:
    print("Nilai C")
else:
    print("Nilai D")
````

## 🔍 6. Operator Perbandingan

Setiap operator perbandingan menghasilkan `True` atau `False` — inilah yang dicek oleh `if`.

| Operator | Arti | Contoh | Hasil |
|:---:|---|---|:---:|
| `==` | Sama dengan | `5 == 5` | `True` |
| `!=` | Tidak sama dengan | `5 != 3` | `True` |
| `>` | Lebih besar | `7 > 2` | `True` |
| `<` | Lebih kecil | `2 < 7` | `True` |
| `>=` | Lebih besar / sama dengan | `5 >= 5` | `True` |
| `<=` | Lebih kecil / sama dengan | `4 <= 3` | `False` |

⚠️ **Jangan tertukar** antara `=` (untuk **menyimpan** nilai) dan `==` (untuk **membandingkan** nilai). Ini adalah salah satu bug paling umum bagi pemula!


````python
# Coba jalankan dan amati hasil tiap perbandingan berikut
print("5 == 5  ->", 5 == 5)
print("5 != 3  ->", 5 != 3)
print("7 > 2   ->", 7 > 2)
print("2 < 7   ->", 2 < 7)
print("5 >= 5  ->", 5 >= 5)
print("4 <= 3  ->", 4 <= 3)
````

### 🧪 Coba Sendiri!
Tebak dulu (di kepala/kertas) hasil dari kode berikut, baru jalankan untuk mengecek jawabanmu.


````python
a = 12
b = 20

print(a == b)
print(a < b)
print(a >= 12)
print(b != 20)
````

## 🔗 7. Operator Logika (`and`, `or`, `not`)

### 📖 Penjelasan
Operator logika memungkinkan kita menggabungkan **beberapa kondisi** sekaligus dalam satu `if`.

| Operator | Arti | Contoh |
|---|---|---|
| `and` | Kedua kondisi harus `True` (irisan) | `umur >= 17 and punya_ktp` |
| `or` | Salah satu kondisi cukup `True` | `hari == "Sabtu" or hari == "Minggu"` |
| `not` | Membalik nilai `True` ↔ `False` | `not sudah_bayar` |


````python
umur = 20
punya_ktp = False

not umur >= 17
````

````python
# Contoh: operator logika
umur = 20
punya_ktp = True

if umur >= 17 and punya_ktp:
    print("Boleh memilih")
else:
    print("Belum bisa memilih")
````

````python
# Contoh: 'or'
hari = "Sabtu"

if hari == "Sabtu" or hari == "Minggu":
    print("Yeay, akhir pekan!")
else:
    print("Masih hari kerja")
````

## 📦 8. Kondisi Bersarang (*Nested Condition*)

### 📖 Penjelasan
`if` di dalam `if` disebut **nested condition**. Kondisi bagian dalam hanya dicek **jika** kondisi bagian luar sudah `True`.

```python
if kondisi_luar:
    if kondisi_dalam:
        ...
    else:
        ...
else:
    ...
```


````python
# Contoh: nested if
umur = 25
member = True

if umur >= 18:
    if member:
        print("Diskon 20% untuk member dewasa")
    else:
        print("Dewasa, tapi belum member")
else:
    print("Belum cukup umur")
````
