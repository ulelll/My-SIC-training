## 1. Rekap Kilat — Variabel & Tipe Data

Variabel adalah "wadah" untuk menyimpan data. Python memiliki beberapa tipe data dasar: `int` (bilangan bulat), `float` (bilangan desimal), `str` (teks), dan `bool` (benar/salah).

```python
nama = "Andi"          # str (teks)
umur = 20               # int (bilangan bulat)
tinggi_badan = 172.5    # float (bilangan desimal)
sudah_lulus = False     # bool (benar/salah)

print(f"Nama: {nama} ({type(nama).__name__})")
print(f"Umur: {umur} ({type(umur).__name__})")
print(f"Tinggi badan: {tinggi_badan} ({type(tinggi_badan).__name__})")
print(f"Sudah lulus: {sudah_lulus} ({type(sudah_lulus).__name__})")
```

---
## 2. Rekap Kilat — Percabangan (If-Else)

Percabangan digunakan untuk membuat program mengambil keputusan berdasarkan suatu kondisi.

```python
nilai = 80

if nilai >= 75:
    status = "Lulus"
else:
    status = "Belum Lulus"

print(f"Nilai {nilai} -> Status: {status}")
```

---
## 3. Rekap Kilat — Perulangan (Loop)

Perulangan (`for` / `while`) digunakan untuk mengulang suatu proses tanpa harus menulis kode yang sama berulang-ulang.

```python
daftar_nilai = [80, 65, 90, 55, 100]

for nilai in daftar_nilai:
    status = "Lulus" if nilai >= 75 else "Belum Lulus"
    print(f"Nilai {nilai} -> {status}")
```

---
## 4. Rekap Kilat — Fungsi

Fungsi digunakan untuk membungkus sekumpulan kode agar bisa dipakai ulang (*reusable*) tanpa menulis ulang logikanya.

```python
def cek_status(nilai):
    # Mengembalikan status kelulusan berdasarkan nilai
    return "Lulus" if nilai >= 75 else "Belum Lulus"

print(cek_status(80))
print(cek_status(60))
```

---
## 5. Rekap Kilat — List & Dictionary

`list` menyimpan kumpulan data berurutan, sedangkan `dictionary` menyimpan data dalam pasangan `key: value` — cocok untuk merepresentasikan satu "baris data" (misalnya satu siswa).

```python
# List: kumpulan nama siswa
daftar_siswa = ["Andi", "Budi", "Citra"]
print("Daftar siswa:", daftar_siswa)

# Dictionary: data nilai per siswa
data_siswa = {"Andi": 80, "Budi": 65, "Citra": 90}

for nama, nilai in data_siswa.items():
    print(f"{nama}: {nilai}")
```

---
## 6. File Handling & CSV

Agar data tidak hilang saat program ditutup, kita perlu menyimpannya ke **file**. Format CSV (*Comma-Separated Values*) adalah format sederhana yang mudah dibaca manusia maupun program, dan bisa dibuka di Excel/Google Sheets.

### 6.1 Menulis data ke CSV

```python
import csv

def simpan_ke_csv(nama, nilai, status, file="nilai_siswa.csv"):
    with open(file, mode="a", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([nama, nilai, status])

# Contoh: menyimpan beberapa data siswa
data_contoh = [
    ("Andi", 80, "Lulus"),
    ("Budi", 65, "Belum Lulus"),
    ("Citra", 90, "Lulus"),
]

for nama, nilai, status in data_contoh:
    simpan_ke_csv(nama, nilai, status)

print("Data berhasil disimpan ke nilai_siswa.csv!")
```

### 6.2 Membaca kembali data dari CSV

```python
import csv

with open("nilai_siswa.csv") as f:
    reader = csv.reader(f)
    data = list(reader)

for baris in data:
    print(baris)
```

## 7. Menggabungkan Semua Konsep

Sekarang kita satukan **fungsi + perulangan + percabangan** menjadi satu program utuh — persis seperti "kepingan puzzle" yang disatukan.

```python
def cek_kelulusan(data_siswa):
    hasil = []
    for nama, nilai in data_siswa.items():   # perulangan (loop)
        if nilai >= 75:                       # percabangan (if-else)
            status = "Lulus"
        else:
            status = "Belum Lulus"
        hasil.append({"nama": nama, "nilai": nilai, "status": status})
    return hasil                              # fungsi mengembalikan nilai

data = {"Andi": 80, "Budi": 65, "Citra": 90, "Dewi": 40}
for siswa in cek_kelulusan(data):
    print(siswa)
```

## 9. Live Build — Membangun Mini App Bertahap

Sekarang kita bangun **Aplikasi Pencatat Nilai Siswa** langkah demi langkah, seperti sesi *live build* di kelas. Setiap langkah akan menulis ulang (*overwrite*) file `app.py`.

### 🔹 Langkah 1: Judul Aplikasi
Kita mulai dengan `st.title()` untuk menampilkan nama aplikasi di bagian paling atas halaman.

```python
%%writefile app.py
import streamlit as st

st.title("Aplikasi Pencatat Nilai Siswa")
st.write("Selamat datang! Silakan isi data siswa di bawah ini.")
```

### 🔹 Langkah 2: Input dari Pengguna
Tambahkan `st.text_input()` dan `st.number_input()` agar pengguna bisa mengisi nama dan nilai siswa.

```python
%%writefile app.py
import streamlit as st

st.title("Aplikasi Pencatat Nilai Siswa")
st.write("Selamat datang! Silakan isi data siswa di bawah ini.")

nama = st.text_input("Nama siswa")
nilai = st.number_input("Nilai", min_value=0, max_value=100, step=1)
```

### 🔹 Langkah 3: Logika (If / Loop)
Tambahkan tombol `st.button()` yang memicu logika `if-else` untuk menentukan status kelulusan.

```python
%%writefile app.py
import streamlit as st

st.title("Aplikasi Pencatat Nilai Siswa")
st.write("Selamat datang! Silakan isi data siswa di bawah ini.")

nama = st.text_input("Nama siswa")
nilai = st.number_input("Nilai", min_value=0, max_value=100, step=1)

if st.button("Simpan"):
    status = "Lulus" if nilai >= 75 else "Belum Lulus"
    st.write(f"Data **{nama}** dengan nilai **{nilai}** -> Status: **{status}**")
```

### 🔹 Langkah 4: Simpan & Tampilkan Data dari CSV
Langkah terakhir: data disimpan ke file CSV, lalu ditampilkan kembali dalam bentuk tabel dengan `st.dataframe()`.

```python
%%writefile app.py
import streamlit as st
import csv
import os

st.title("Aplikasi Pencatat Nilai Siswa")
st.write("Selamat datang! Silakan isi data siswa di bawah ini.")

nama = st.text_input("Nama siswa")
nilai = st.number_input("Nilai", min_value=0, max_value=100, step=1)

FILE_CSV = "nilai_siswa_app.csv"

if st.button("Simpan"):
    status = "Lulus" if nilai >= 75 else "Belum Lulus"
    with open(FILE_CSV, mode="a", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([nama, nilai, status])
    st.success(f"Data {nama} berhasil disimpan dengan status: {status}")

st.subheader("Riwayat Data Siswa")
if os.path.exists(FILE_CSV):
    with open(FILE_CSV) as f:
        data = list(csv.reader(f))
    st.dataframe(data, use_container_width=True)
else:
    st.info("Belum ada data yang disimpan.")
```
