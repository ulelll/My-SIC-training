## 2️⃣ Membuka File di Python: Mode Akses

Setiap kali membuka file, kita perlu menentukan **mode**:

| Mode | Arti | Kegunaan |
|------|------|----------|
| `"r"` | *read* | Membuka file untuk **dibaca saja** |
| `"w"` | *write* | Membuka untuk **menulis** — isi lama akan **ditimpa** (hilang!) |
| `"a"` | *append* | Membuka untuk **menambah** data di baris paling akhir |

### Perintah `open()` dan pentingnya `close()`

Setiap kali kita `open()` sebuah file, kita **wajib** menutupnya kembali dengan `close()`. Kalau lupa, file bisa "terkunci" atau menyebabkan pemborosan memori.

````python
# Membuka file dengan mode "w" untuk menulis
file = open("data_latihan/data.txt", "r")
isi_awal = file.read()
file.close()  # wajib! menutup pintu file

print(isi_awal)
````

## 3️⃣ Membaca File: `read()`, `readline()`, `readlines()`

Ada tiga cara umum untuk membaca isi file:

- **`read()`** → mengambil **seluruh isi** file sebagai satu teks panjang
- **`readline()`** → mengambil **satu baris saja**, dari atas ke bawah
- **`readlines()`** → mengambil **semua baris**, hasilnya berupa **list** of string

Mari kita coba satu per satu. Pertama, buat dulu file dengan beberapa baris.

````python
with open("data_latihan/catatan_harian.txt", "w") as file:
    file.write("Bangun pagi\n")
    file.write("Belajar Python\n")
    file.write("Mengerjakan tugas\n")

print("File catatan_harian.txt berhasil dibuat ✅")
````

## 5️⃣ Menulis ke File: Mode `"w"` vs `"a"`

- Mode **`"w"`** akan **menimpa** seluruh isi file yang lama
- Mode **`"a"`** akan **menambahkan** baris baru di akhir, isi lama tetap aman

Mari kita buktikan perbedaannya.

````python
# Mode "w" -> menimpa isi lama
with open("data_latihan/log.txt", "w") as file:
    file.write("Baris A")

with open("data_latihan/log.txt", "r") as file:
    print("Isi setelah mode w pertama kali:", file.read())
````

````python
# Menulis LAGI dengan mode "w" -> isi lama HILANG, tertimpa total
with open("data_latihan/log.txt", "w") as file:
    file.write("Baris B")

with open("data_latihan/log.txt", "r") as file:
    print("Isi setelah mode w kedua kali:", file.read())
    print("⚠️  Perhatikan: 'Baris A' sudah hilang, tertimpa oleh 'Baris B'")
````

````python
# Mode "a" -> menambah tanpa menghapus isi lama
with open("data_latihan/log.txt", "a") as file:
    file.write("\nBaris tambahan di akhir")

with open("data_latihan/log.txt", "r") as file:
    print("Isi setelah mode a:")
    print(file.read())
    print("✅ 'Baris B' masih ada, dan baris baru ditambahkan di bawahnya")
````

## 6️⃣ Apa itu CSV?

**CSV (Comma-Separated Values)** adalah cara menyimpan data tabel (seperti spreadsheet) dalam bentuk **teks murni**, di mana setiap kolom dipisahkan dengan tanda koma (`,`).

Contoh tabel:

| nama | usia | kota |
|------|------|------|
| Rani | 17 | Bandung |
| Budi | 16 | Jakarta |
| Sari | 17 | Surabaya |

Kalau ditulis dalam format CSV, jadinya seperti ini:

```
nama,usia,kota
Rani,17,Bandung
Budi,16,Jakarta
Sari,17,Surabaya
```

Baris pertama biasanya disebut **header** (nama kolom), dan baris-baris setelahnya adalah **data**.

Mari kita buat file CSV ini secara langsung dari Python.

````python
with open("data_latihan/siswa.csv", "w") as file:
    file.write("nama,usia,kota\n")
    file.write("Rani,17,Bandung\n")
    file.write("Budi,16,Jakarta\n")
    file.write("Sari,17,Surabaya\n")

print("File siswa.csv berhasil dibuat ✅")

with open("data_latihan/siswa.csv", "r") as file:
    print(file.read())
````

## 7️⃣ Membaca CSV dengan `split()`

Karena CSV hanyalah teks biasa dengan koma sebagai pemisah, kita bisa memecah setiap baris menjadi list menggunakan `split(",")`.

Alurnya:

```
"Rani,17,Bandung"  --split(",")-->  ["Rani", "17", "Bandung"]
```

> 💡 Kita juga memakai `.strip()` sebelum `.split()` untuk menghapus karakter enter (`\n`) di akhir baris.

````python
with open("data_latihan/siswa.csv", "r") as file:
    header = file.readline()  # lewati baris pertama (header)
    print("Header:", header.strip())
    print("---")
    for baris in file:
        data = baris.strip().split(",")
        print(data)
````

````python
# Mengambil kolom tertentu saja, misalnya hanya nama dan kota
with open("data_latihan/siswa.csv", "r") as file:
    file.readline()  # lewati header
    for baris in file:
        data = baris.strip().split(",")
        nama = data[0]
        usia = data[1]
        kota = data[2]
        print(f"{nama} usia {usia} tinggal di {kota}")
````

### 📋 Tugas 2: Membuat & Membaca CSV Sendiri

Buatlah file CSV bernama `data_latihan/produk.csv` berisi data minimal 4 produk dengan kolom: `nama_produk`, `harga`, `stok`.

Lalu tulis program untuk:
1. Membaca file CSV tersebut baris per baris (**tanpa Pandas**, gunakan `split()`)
2. Menampilkan setiap produk dengan format: `Nama Produk - Rp Harga (Stok: X)`
3. Menghitung dan menampilkan **total nilai stok** (harga × stok) dari seluruh produk

````python
# TUGAS 2 - Tulis kodemu di sini

with open("data_latihan/produk.csv", "w") as file:
    file.write("nama_produk,harga,stok\n")
    file.write("Buku,25000,10\n")
    file.write("Pulpen,5000,20\n")
    file.write("Pensil,3000,15\n")
    file.write("Penghapus,2000,10\n")
total_nilai_stok = 0

with open("data_latihan/produk.csv", "r") as file:
    file.readline()

    for baris in file:
        data = baris.strip().split(",")
        nama_produk = data[0]
        harga = int(data[1])
        stok = int(data[2])

        print(f"{nama_produk} - Rp {harga} (Stok: {stok})")
        total_nilai_stok += harga * stok

print(f"Total nilai stok: Rp {total_nilai_stok}")
````
